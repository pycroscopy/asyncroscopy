## MCP server usecase on twin

## Prereqs

Start the Tango stack on the microscope computer. Start MCP and the LLM device on the LLM computer. Both MCP and the LLM device use the microscope computer's Tango database:

```bash
uv run startup_scripts/run_servers.py --yaml configs/DigitalTwin.yaml
uv run startup_scripts/run_mcp.py --yaml configs/mcp_dt.yaml

### Connect to MCP server

````python
from fastmcp import Client

MCP_URL = "http://localhost:8000/mcp"

client = Client(MCP_URL)

import os
````

### List Tools of MCP server

````python
async with client:
    tools = await client.list_tools()

for tool in tools:
    print(tool.name)
````

````python
# Acquire data using MCP
async with Client(MCP_URL) as client:
    result = await client.call_tool(
        "DigitalTwin_acquire_camera_image",
        {}
    )

key = result.structured_content["result"]
key
````

````python
# Get data from key
async with Client(MCP_URL) as client:
    image_info = await client.call_tool(
        "get_data_from_key",
        {"key": key}
    )

image_info.data.keys()
````
